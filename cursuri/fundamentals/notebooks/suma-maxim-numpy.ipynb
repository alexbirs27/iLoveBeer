# Suma si maximul, cu NumPy

Aceleasi doua exercitii din Curs 0, de data asta fara bucla scrisa de
mana. Compara cu `suma()` si `maxim()` de acolo, e acelasi calcul.

```python
import numpy as np

v = np.array([3, 7, 2, 9, 4])

v.sum()   # 25, acelasi rezultat ca suma() scrisa de mana
v.max()   # 9, acelasi rezultat ca maxim() scrisa de mana
```
