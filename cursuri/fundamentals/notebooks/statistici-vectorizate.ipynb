# Statistici fara bucla

Aceeasi medie si deviatie standard, calculate intai cu `for` (ca sa vedem
exact ce calculeaza), apoi vectorizat. Comparam si timpul de rulare pe
un vector mare.

```python
import numpy as np

v = np.random.rand(1_000_000)
```

```python
# varianta cu bucla — asta calculeaza de fapt v.mean()
def medie_bucla(v):
    s = 0.0
    for x in v:
        s += x
    return s / len(v)

%timeit medie_bucla(v)
```

```python
# varianta vectorizata
%timeit v.mean()
```

Diferenta de viteza nu vine din faptul ca NumPy calculeaza altceva — e
aceeasi suma impartita la n. Vine din faptul ca `v.mean()` ruleaza intr-o
bucla scrisa in C, peste memorie continua, fara overhead-ul obiectelor
Python boxate din `medie_bucla`.

## Normalizare (z-score)

$z = \dfrac{x - \mu}{\sigma}$ — aducem datele la medie 0, deviatie 1.

```python
z = (v - v.mean()) / v.std()

z.mean(), z.std()  # ~0, ~1
```
