## Variabile si tipuri

Nu declari tipul, Python il deduce din valoare. Afli tipul cu `type()`, si convertesti intre tipuri cu `int()`, `float()`, `str()`.

Conversiile astea apar constant cand citesti date brute - dintr-un fisier, de la tastatura, mai tarziu dintr-un CSV cu Pandas - inainte sa le prelucrezi ca numere.

O variabila e o referinta spre un obiect, nu o cutie cu valoarea inauntru. `y = x` copiaza referinta, nu valoarea - conteaza mai ales la liste, o sa vezi mai jos de ce. Exista si `None`, valoarea "fara valoare" - implicita pentru o functie care nu returneaza nimic, sau pentru un parametru optional.

````python
n = 5           # int
x = 3.14        # float
nume = 'Alex'   # str
gasit = True    # bool

print(type(n), type(x))

# conversii - des intalnite cand citesti input ca text si vrei numar
text = '42'
numar = int(text)
print(numar + 1)

a = 2500
b = a            # b e aceeasi referinta ca a, nu o copie
print(a is b)    # True, acelasi obiect in memorie

lipsa = None
print(bool(0), bool(''), bool([]), bool(lipsa))   # toate False
````

## Operatori

Aritmeticii sunt aceiasi ca in C++, cu doua exceptii: `/` da mereu rezultat zecimal, chiar intre doi intregi, si `**` e ridicare la putere. Pentru impartire intreaga foloseste `//`. Operatorii logici sunt cuvinte, nu simboluri: `and`, `or`, `not` in loc de `&&`, `||`, `!`.

Toate astea sunt O(1), cost constant. Cand o sa faci acelasi calcul pe un vector intreg cu NumPy, in loc de un numar, exact acesti operatori se aplica element cu element - doar mult mai rapid decat daca ai scrie tu bucla in Python.

Python nu are `++`/`--`. In schimb are un operator ternar: `valoare_daca_adevarat if conditie else valoare_daca_fals`, echivalentul lui `conditie ? a : b` din C++.

**Doua capcane clasice:** `is` verifica daca doua variabile sunt *acelasi obiect* in memorie, nu daca au aceeasi valoare - pentru valori, foloseste `==`. Si la numere zecimale, `==` poate da rezultate surprinzatoare din cauza reprezentarii in virgula mobila - compara cu o toleranta, nu cu egalitate stricta.

````python
print(7 / 2)    # 3.5, nu 3
print(7 // 2)   # 3, impartire intreaga
print(7 % 2)    # 1, rest
print(2 ** 10)  # 1024

a, b = 5, 10
print(a < b and b < 100)

n = -3
paritate = 'par' if n % 2 == 0 else 'impar'
print(paritate)

v1 = [1, 2, 3]
v2 = [1, 2, 3]
print(v1 == v2)   # True, aceeasi valoare
print(v1 is v2)   # False, obiecte diferite in memorie

print(0.1 + 0.2 == 0.3)             # False! eroare de reprezentare in virgula mobila
print(abs(0.1 + 0.2 - 0.3) < 1e-9)  # asa compari corect doua flotante
````

## Stringuri

Un string se comporta ca o lista de caractere: il indexezi si faci slicing pe el, la fel. Metode pe care le folosesti des: `split`, `strip`, `join`, `lower`/`upper`, `replace`. Pentru afisare cu valori inauntru, foloseste f-strings - mai lizibil decat concatenare cu `+`.

**Complexitate:** indexare si `len()` sunt O(1). Slicing si `split()` sunt O(n), fac o copie noua. Daca concatenezi stringuri cu `+` intr-o bucla, fiecare `+` creeaza un string nou, deci tot procesul ajunge O(n^2) - pentru multe concatenari, aduna bucatile intr-o lista si fa `''.join(lista)` la final, e O(n).

Iti trebuie la orice procesare de text - loguri, fisiere, mai tarziu feature-uri extrase din text pentru un model.

Un string e imutabil - nicio metoda nu il modifica pe loc, toate returneaza un string nou. Daca faci `s.upper()` fara sa retii rezultatul, `s` ramane neschimbat. Din acelasi motiv, Python poate refolosi acelasi obiect pentru stringuri scurte identice - de-aia comparatia `is` pe stringuri e nesigura (poate fi `False` chiar daca valorile sunt egale). Foloseste mereu `==` pentru continut, nu `is`.

````python
s = 'Ana are mere'

print(s[0])       # 'A'
print(s[-1])      # ultimul caracter
print(s[4:8])     # 'are '

cuvinte = s.split(' ')      # ['Ana', 'are', 'mere']
print(cuvinte)

n = 5
print(f'avem {n} mere')     # f-string, mai lizibil decat 'avem ' + str(n) + ' mere'

s.upper()            # nu modifica s, rezultatul se pierde daca nu il retii
print(s)              # tot 'Ana are mere'

s_mare = s.upper()    # asa retii rezultatul
print(s_mare)
````

## Liste

`list`-ul e vectorul din C++, dar tine orice tip amestecat si isi schimba singur dimensiunea. Slicing-ul merge la fel ca la stringuri. `sorted()` face o copie sortata, `.sort()` sorteaza pe loc; ambele accepta `key` daca vrei alt criteriu decat cel implicit.

**Complexitate:** acces prin index si `.append()` sunt O(1). `.insert()` sau `.remove()` la inceput sau la mijloc sunt O(n), pentru ca muta tot ce urmeaza dupa punctul ala. Cautarea `x in v` e O(n), parcurge lista pana gaseste (sau pana termina). `sorted()`/`.sort()` sunt O(n log n).

NumPy inlocuieste `list`-ul cu un array de memorie contigua si tipat - o sa vezi in capitolul urmator ca de-aia operatiile pe el sunt mult mai rapide decat pe un `list` obisnuit.

`.extend(alta_secventa)` adauga fiecare element din `alta_secventa` in `v`, unul cate unul - spre deosebire de `.append()`, care adauga `alta_secventa` intreaga, ca un singur element nou.

````python
v = [3, 1, 4, 1, 5, 9]

v.append(2)
print(v)
print(v[1:4])          # slicing, ca la stringuri
print(sorted(v))
print(sorted(v, key=lambda x: -x))   # descrescator

for i, val in enumerate(v):    # index + valoare, deodata
    print(i, val)

v.extend([7, 8])     # adauga 7 si 8, separat
print(v)

v.append([7, 8])     # adauga [7, 8] ca UN singur element
print(v)
````

## Matrici (liste 2D) - capcana referintelor comune

O matrice in Python e o lista de liste. Dar `[[0] * n] * m` **nu** creeaza `m` randuri independente - creeaza `m` referinte catre *aceeasi* lista. Modifici un rand, se modifica toate. Capcana clasica pentru cineva venit din C++, unde `vector<vector<int>>` nu are problema asta.

````python
gresit = [[0] * 3] * 3    # 3 referinte catre ACEEASI lista interioara
gresit[1][1] = 7
print(gresit)              # 7 apare pe toate cele 3 randuri!

corect = [[0] * 3 for _ in range(3)]   # comprehension - o lista noua de fiecare data
corect[1][1] = 7
print(corect)                            # doar randul 1 s-a schimbat
````

## Copiere: referinta, shallow copy, deep copy

Trei niveluri, si conteaza care il alegi:

- `b = a` - copiaza doar referinta. `a` si `b` sunt *acelasi* obiect, modifici unul, se modifica si celalalt.
- `b = a.copy()` (sau `a[:]`) - **shallow copy**. Copiaza lista de nivelul 1, dar daca elementele sunt la randul lor mutabile (ex: liste in liste), acelea raman partajate.
- `copy.deepcopy(a)` - copiaza totul, recursiv, pana la capat. Corect intotdeauna, dar poate fi lent - foloseste-l doar cand chiar ai nevoie de o copie complet independenta.

````python
import copy

a = [1, [2, 3]]

b_ref = a                    # aceeasi referinta
b_shallow = a.copy()         # shallow copy
b_deep = copy.deepcopy(a)    # deep copy

a[0] = 100          # schimba un element simplu
a[1][0] = 999       # schimba un element DIN LISTA INTERIOARA

print(b_ref)        # [100, [999, 3]] - e literalmente a
print(b_shallow)    # [1, [999, 3]]   - nivelul 1 e independent, dar lista interioara e tot partajata
print(b_deep)        # [1, [2, 3]]     - complet independenta
````

## List comprehension

O bucla scurta, scrisa pe o linie, ca sa construiesti o lista noua dintr-una veche. E echivalentul lui:

```
rezultat = []
for x in v:
    if conditie(x):
        rezultat.append(expresie(x))
```

scris compact ca `[expresie(x) for x in v if conditie(x)]`.

**Complexitate:** aceeasi ca bucla explicita, O(n) - nu e mai rapida ca algoritm, doar mai rapida de scris si de citit. O sa dai peste tiparul asta des la preprocesare de date, inainte sa ajungi la NumPy/Pandas.

````python
v = [1, 2, 3, 4, 5, 6]

patrate = [x * x for x in v]
print(patrate)

pare = [x for x in v if x % 2 == 0]
print(pare)
````

## Tupluri

Ca o lista, dar imutabila - o data creat, nu-i mai schimbi elementele. Le folosesti mai ales cand o functie trebuie sa returneze mai multe valori deodata.

O sa le intalnesti des ca valoare de return: `train_test_split()` din scikit-learn, de exemplu, returneaza un tuplu de 4 liste deodata.

Un tuplu cu un singur element are nevoie de virgula: `(5,)`, nu `(5)` - fara virgula e doar numarul 5 intre paranteze. La despachetare, `*` string mai multe elemente intr-o lista: `prim, *rest = [1, 2, 3, 4]` da `prim = 1`, `rest = [2, 3, 4]`.

````python
punct = (3, 4)
x, y = punct     # unpacking
print(x, y)

def minim_si_maxim(v):
    return min(v), max(v)   # returneaza un tuplu

mn, mx = minim_si_maxim([5, 2, 8, 1])
print(mn, mx)

unu = (5,)
print(type(unu))     # tuple, nu int

prim, *rest = [1, 2, 3, 4]
print(prim, rest)
````

## Dictionare

Cheie -> valoare, cautare rapida fara sa parcurgi toata structura. `.get(cheie, implicit)` iti da o valoare implicita daca cheia nu exista, in loc sa arunce eroare. Iterezi cu `.items()` cand vrei si cheia si valoarea.

**Complexitate:** acces, inserare si stergere sunt O(1) in medie - Python foloseste un hash table pe dedesubt, nu parcurge cheile una cate una. De-aia un dictionar e mult mai rapid decat o lista cand cauti dupa cheie.

Un DataFrame din Pandas e conceptual un dictionar de coloane (nume coloana -> vector de date) - o sa recunosti tiparul cand ajungem acolo.

De la Python 3.7, un dictionar isi tine minte ordinea in care ai adaugat cheile. `.keys()`, `.values()`, `.items()` sunt "ferestre" vii spre dictionar, se actualizeaza automat daca il modifici. `.pop(cheie, implicit)` sterge si returneaza valoarea, cu acelasi fallback ca la `.get()`.

````python
varste = {'Ana': 20, 'Bogdan': 22} # ce exista cand definesti dictionarul

varste['Carla'] = 19            # adaugi o cheie noua
print(varste.get('Dan', 0))     # 0, cheia nu exista

for nume, varsta in varste.items():
    print(nume, varsta)

varsta_scoasa = varste.pop('Ana', 0)
print(varsta_scoasa, varste)
````

## Cum primesc functiile argumentele

Nu e nici "prin valoare", nici "prin referinta" ca in C++ - e "prin obiect". Functia primeste o referinta catre acelasi obiect. Ce se intampla dupa depinde ce faci cu ea:

- daca *muti obiectul* (ex: `lista.append(x)`) - schimbarea se vede si in afara functiei, pentru ca e acelasi obiect.
- daca *reatribui parametrul* (ex: `lista = lista + [x]`) - creezi un obiect nou, local; in afara functiei, variabila originala nu se schimba.

````python
def adauga_in_loc(lista, x):
    lista.append(x)    # muta obiectul primit - vizibil in afara

def adauga_nou(lista, x):
    lista = lista + [x]    # reatribuie parametrul local - NU se vede in afara

v = [1, 2, 3]
adauga_in_loc(v, 4)
print(v)     # [1, 2, 3, 4] - s-a schimbat

adauga_nou(v, 5)
print(v)     # tot [1, 2, 3, 4] - neschimbat, 5 s-a pierdut cu functia
````

## for...else si while...else

Ceva ce C++ nu are: un `for` sau `while` poate avea un `else`, care se executa doar daca bucla s-a terminat **fara** `break`. Util cand cauti ceva si vrei sa stii daca l-ai gasit sau ai parcurs tot degeaba - fara sa mai ai nevoie de un steag separat (`gasit = True/False`).

````python
def cel_mai_mic_prim(a, b):
    for n in range(a, b + 1):
        for d in range(2, n):
            if n % d == 0:
                break     # n nu e prim, treci la urmatorul n
        else:
            return n      # bucla interioara s-a terminat FARA break -> n e prim
    return None           # niciun prim gasit in [a, b]

print(cel_mai_mic_prim(8, 20))   # 11
````

## Citire de fisiere

`open()` cu `with` inchide automat fisierul la final, chiar daca apare o eroare pe parcurs. Citesti linie cu linie, iterand direct peste fisier. Refolosim `exemplu-note.csv` din playground-ul de test.

**Complexitate:** citirea e O(n) in dimensiunea fisierului. Citind linie cu linie, ca mai jos, nu tii tot fisierul in memorie deodata - conteaza cand fisierul e mare. `pd.read_csv()`, din capitolul urmator, face exact asta pe dedesubt, doar mult mai eficient.

Modul de deschidere e al doilea argument: `"r"` (citire, implicit), `"w"` (scriere - **sterge tot ce era in fisier**), `"a"` (adauga la final, fara sa stearga), `"x"` (creeaza, eroare daca fisierul exista deja). La scriere, `.write()` nu adauga `\n` singur - il pui tu explicit.

````python
def fabrica_inmultitor(k):
    def inmultitor(x):
        return x * k    # 'k' vine din functia parinte, e "retinut" aici
    return inmultitor

dublu = fabrica_inmultitor(2)
triplu = fabrica_inmultitor(3)
print(dublu(5), triplu(5))    # 10 15

studenti = [('Ana', 9.5), ('Bogdan', 7.2), ('Carla', 8.8)]
print(sorted(studenti, key=lambda s: s[1], reverse=True))   # descrescator dupa nota
````

## Generatori

O functie normala calculeaza tot si returneaza tot deodata. Un generator (foloseste `yield` in loc de `return`) produce valorile una cate una, la cerere - nu tine toata secventa in memorie deodata. Util cand lucrezi cu multe date si nu vrei sa le incarci pe toate simultan.

````python
def patrate_pana_la(n):
    for x in range(n):
        yield x * x    # 'pauzeaza' aici, reia de unde a ramas la urmatorul next()

for p in patrate_pana_la(5):
    print(p)

g = patrate_pana_la(3)
print(next(g))    # 0
print(next(g))    # 1 - a continuat de unde a ramas
````

Asta e tot ce-ti trebuie ca sa incepi. Restul il inveti pe masura ce dai peste el, rezolvand probleme - exact ca la C++. Urmeaza notebook-ul de exercitii, unde le pui in practica singur, si dupa el capitolul de tehnici algoritmice (greedy, divide et impera, backtracking, programare dinamica) - acolo recursivitatea si complexitatea de mai sus devin uneltele principale de lucru.

````python
def factorial(n):
    if n == 0:           # caz de baza - fara el, recursivitate infinita
        return 1
    return n * factorial(n - 1)

print(factorial(5))    # 120

import sys
print(sys.getrecursionlimit())   # 1000, implicit

# Fibonacci recursiv naiv: recalculeaza acelasi subrezultat de multe ori - O(2^n)
def fib_naiv(n):
    if n <= 1:
        return n
    return fib_naiv(n - 1) + fib_naiv(n - 2)

# Fibonacci iterativ: fiecare valoare se calculeaza o singura data - O(n)
def fib_iterativ(n):
    a, b = 0, 1
    for _ in range(n):
        a, b = b, a + b
    return a

print(fib_naiv(20), fib_iterativ(20))   # acelasi rezultat, viteza foarte diferita
````

## Complexitatea algoritmilor, mai in detaliu

Cateva reguli si repere care iti raman utile toata cariera, nu doar la cursul asta.

**Simplificare:** la n mare, constantele nu conteaza - `O(5n + 3)` e `O(n)`, `O(3n^2 + 5n)` e `O(n^2)`. Pastrezi doar termenul care creste cel mai repede.

**Compunere:** o bucla de n ori cu un bloc O(1) inauntru da O(n). O bucla de n ori cu un bloc O(m) inauntru da O(n*m). Doua bucle *secventiale* (una dupa alta, nu una in alta) se aduna: O(n + m). Doua bucle *imbricate* se inmultesc: O(n * m).

**Clase de complexitate, de la cea mai rapida la cea mai lenta** (cu exemple din ce ai vazut deja):

| Clasa | Exemplu |
|---|---|
| O(1) | acces prin index, `.append()` |
| O(log n) | cautare binara intr-o lista sortata |
| O(n) | o parcurgere, `x in lista`, suma cifrelor unui numar |
| O(n log n) | `sorted()`, majoritatea algoritmilor de sortare |
| O(n^2) | doua bucle imbricate, sortare naiva |
| O(2^n) | generarea tuturor submultimilor unei multimi |
| O(n!) | generarea tuturor permutarilor |

Diferenta dintre polinomial (`n^2`) si exponential (`2^n`) nu e doar teoretica - la 500.000 de operatii/secunda, generarea submultimilor pentru n=30 dureaza ~35 minute; pentru n=50, ~72 de ani; pentru n=100, un numar de ani mai mare decat varsta universului de multe ori. De-aia in capitolul urmator o sa vezi mereu intrebarea "exista o varianta mai rapida?" inainte sa acceptam un algoritm exponential.

**Atentie la variabile care se dubleaza in interiorul unei bucle** - schimba complet clasa de complexitate, chiar daca bucla "arata" liniara.

````python
# Arata O(n) la prima vedere: o bucla exterioara de n ori.
# Dar p se dubleaza de fiecare data - bucla interioara face 2^n pasi in total, nu n.
def capcana_dublare(n):
    pasi = 0
    p = 1
    for i in range(n):
        j = 0
        while j < p:
            pasi += 1
            j += 1
        p *= 2
    return pasi

print(capcana_dublare(10))    # 2^10 - 1 = 1023 pasi, nu ~10

# Two pointers: doi indici care se apropie unul de altul, fiecare avanseaza cel mult o data.
# Gaseste doua numere dintr-o lista SORTATA care insumeaza 'target'.
def doua_numere(v, target):
    stanga, dreapta = 0, len(v) - 1
    while stanga < dreapta:
        s = v[stanga] + v[dreapta]
        if s == target:
            return v[stanga], v[dreapta]
        elif s < target:
            stanga += 1
        else:
            dreapta -= 1
    return None

print(doua_numere([1, 3, 4, 6, 8, 11], 10))   # (4, 6), O(n) - nu O(n^2) ca varianta cu doua bucle
````

## Seturi

O colectie fara duplicate si fara ordine. Utile cand te intereseaza doar ce elemente apar, nu de cate ori sau in ce ordine, sau cand faci operatii de multimi.

**Complexitate:** la fel ca la dictionar, `in`, `.add()` si `.remove()` sunt O(1) in medie - un `set` e tot un hash table pe dedesubt, doar fara valori atasate cheilor. Cautarea `x in set` e mult mai rapida decat `x in lista`.

Util la eliminat duplicate rapid, sau la verificat daca un cuvant apare intr-un vocabular, la procesare de text.

Poti construi un set si prin comprehension: `{x for x in v}`. Operatorul `<=` intre doua seturi testeaza submultime, nu ordine. `.add(x)` adauga un singur element - daca `x` e el insusi o colectie, `.update(x)` ii adauga elementele unul cate unul, nu pe `x` ca bloc. `.discard(x)` nu arunca eroare daca `x` lipseste, `.remove(x)` da.

````python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a & b)   # intersectie: {3, 4}
print(a | b)   # reuniune
print(a - b)   # diferenta: {1, 2}

v = [1, 2, 2, 3, 3, 3]
print(set(v))  # elimina duplicatele

c = {x for x in range(10) if x % 3 == 0}
print(c)

print({1, 2} <= {1, 2, 3})   # True, submultime

s = {1, 2, 3}
s.add((9, 10))        # adauga tuplul (9, 10) ca UN element
s.update([9, 10])     # adauga 9 si 10, separat
print(s)

s.discard(999)   # nu exista in s, dar nu da eroare
````

## Functii

`def` cu argumente care pot avea valoare implicita. Daca nu stii dinainte cate argumente primesti, `*args` le string pe toate intr-un tuplu.

Orice model din scikit-learn sau PyTorch e, pe dedesubt, o gramada de functii ca astea - `.fit()`, `.predict()` - doar atasate unui obiect. Acelasi `def`, aceleasi argumente cu valoare implicita pe care le vezi aici.

Mai jos intram putin mai in detaliu: cum primesc de fapt functiile argumentele, functii care tin minte context (closures), si functii "lenese" (generatori).

````python
def putere(x, n=2):     # n are valoare implicita
    return x ** n

print(putere(3))        # 9, foloseste n=2
print(putere(3, 3))     # 27

def suma_toate(*numere):
    return sum(numere)

print(suma_toate(1, 2, 3, 4))
````

## while, break, continue

Ca in C++. `break` iese din bucla, `continue` sare direct la urmatoarea iteratie, fara sa mai execute restul corpului.

````python
i = 0
while i < 10:
    i += 1
    if i % 2 == 0:
        continue      # sare peste pare
    if i > 7:
        break
    print(i)
````

## Gestionarea erorilor

`try`/`except` prinde o eroare in loc sa opreasca tot programul. Util mai ales la citirea de input, unde nu controlezi ce tasteaza utilizatorul.

O sa-ti trebuiasca des cand incarci date reale - un CSV cu o valoare lipsa sau un rand corupt iti poate crapa tot scriptul daca nu prinzi eroarea la timp.

````python
def citeste_numar(text):
    try:
        return int(text)
    except ValueError:
        print(f"'{text}' nu e un numar")
        return None

print(citeste_numar('42'))
print(citeste_numar('abc'))
````

## Citire de fisiere

`open()` cu `with` inchide automat fisierul la final, chiar daca apare o eroare pe parcurs. Citesti linie cu linie, iterand direct peste fisier. Refolosim `exemplu-note.csv` din playground-ul de test.

**Complexitate:** citirea e O(n) in dimensiunea fisierului. Citind linie cu linie, ca mai jos, nu tii tot fisierul in memorie deodata - conteaza cand fisierul e mare. `pd.read_csv()`, din capitolul urmator, face exact asta pe dedesubt, doar mult mai eficient.

````python
!wget -q https://raw.githubusercontent.com/alexbirs27/iLoveBeer/main/cursuri/data/exemplu-note.csv

with open('exemplu-note.csv') as f:
    for linie in f:
        print(linie.strip())    # strip scoate \n de la capatul liniei
````

## Cateva module utile din biblioteca standard

`math` pentru functii matematice, `random` pentru numere aleatoare, `collections.Counter` pentru numarat aparitii intr-o singura linie, in loc sa scrii singur bucla cu dictionar.

`Counter` e tot un dictionar pe dedesubt, deci acelasi cost O(1) per operatie ca la sectiunea de dictionare de mai sus.

````python
import math
print(math.sqrt(16), math.ceil(4.1), math.floor(4.9))

import random
print(random.randint(1, 6))       # arunca un zar
print(random.choice(['a', 'b', 'c']))

from collections import Counter
cuvinte = 'ana are mere ana are pere'.split()
print(Counter(cuvinte))            # cate aparitii are fiecare cuvant
````
