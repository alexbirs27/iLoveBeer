# Maximul dintr-un vector

Retinem cel mai mare element vazut pana acum, pe masura ce parcurgem
lista. Pornim de la primul element, nu de la 0 sau minus infinit, ca sa
mearga si pentru vectori cu toate elementele negative.

```python
def maxim(v):
    m = v[0]
    for x in v[1:]:
        if x > m:
            m = x
    return m

maxim([3, 7, 2, 9, 4])  # 9
```
