## De mana

Media e suma impartita la cate elemente sunt

```python
import csv

note = []
with open('exemplu-note.csv') as f:
    for rand in csv.DictReader(f):
        note.append(float(rand['nota']))

media = sum(note) / len(note)
print(media)
```
